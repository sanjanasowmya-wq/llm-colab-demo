types of codes available for the array module for various data types:
1. "i">integer type-2
2. "f">float type-4
3. "d">float type-8
4. "u">unicode character type-2

```python
import array as myarray

```

```python
arr0=myarray.array("i",[1,2,3])
arr1=myarray.array("d",[1.0,2.3,4.5])
arr2=myarray.array("u",["a","b","c","d"])
#typecodes
print(arr0.typecode)#i
print(arr1.typecode)#d
print(arr2.typecode)#u
print(arr0)
print(arr1)
print(arr2)
```

```python
#printing the length of the array:
print(len(arr0))
print(len(arr1))
print(len(arr2))
```

```python
for i in range(0,len(arr0)):
  print(arr0[i],end=' ')
print("\n")
for i in range(0,len(arr1)):
  print(arr1[i],end=' ')
print("\n")
for i in range(0,len(arr2)):
  print(arr2[i],end=' ')
print("\n")
```

///This loop iterates through the elements of the array arr0.

///The range(0, len(arr0)) function generates a sequence of indices from 0 to len(arr0) - 1, inclusive.

///It represents the valid index range for accessing elements of arr0.
///Inside the loop, arr0[i] accesses the element at index i of arr0, and print(arr0[i], end=' ') prints that element followed by a space (using the end=' ' argument of the print function).

///After printing all elements of arr0, print("\n") prints a newline character to move to the next line for clarity.


INDEXING

```python
x=list(range(1,100))
arr=myarray.array("i",x)
for i in range(0,len(arr)):
  print(arr[i],end=" ")
print("\n")
print(arr[25])#indexing
```

 **Adding elements in the array by insert and append methods**

```python
import array as myarray
new_arr=myarray.array("i",[3,4,5])
for i in range(0,len(new_arr)):
  print(new_arr[i],end=" ")
print("\n")
new_arr.insert(0,2)
new_arr.insert(0,1)
for i in range(0,len(new_arr)):
  print(new_arr[i],end=" ")
print("\n")
new_arr.append(6)
for i in range(0,len(new_arr)):
  print(new_arr[i],end=" ")
print("\n")
```

**Updating elements in an array**

```python
import array as myarr
newarr=myarr.array("i",[1,2,2,4,5])
for i in range(0,len(newarr)):
  print(newarr[i],end=" ")
print("\n")
newarr[2]=3
for i in range(0,len(newarr)):
  print(newarr[i],end=" ")
print("\n")
```

```python
def alternatenums(arr):
  for i in range(0,len(arr),2):
    print(arr[i],end=" ")
arr1=[1,2,3,4,5,6,7,9,10]
arr2=[10,20,30,40,50,60,70,80,90]
alternatenums(arr1)
print("\n")
alternatenums(arr2)
print("\n")
```

```python
import array as myarray
newarr=myarray.array("i",[1,2,3,4,5,6])
for i in range(0,len(newarr)):
  print(newarr[i],end=" ")
print("\n")
newarr.pop(5)
for i in range(0,len(newarr)):
  print(newarr[i],end=" ")
print("\n")
newarr.remove(5)
for i in range(0,len(newarr)):
  print(newarr[i],end=" ")
print("\n")
```
